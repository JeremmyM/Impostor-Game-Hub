```python
# ==============================================================================
# Разработка системы автоматического контроля СИЗ с применением компьютерного зрения
# ==============================================================================

# 1. Подготовка окружения и библиотек
!pip install -U ultralytics kagglehub pandas matplotlib

import os
import time
import yaml
import torch
import pandas as pd
import matplotlib.pyplot as plt
from ultralytics import YOLO, RTDETR
import kagglehub
from google.colab import drive
```

```python
# 2. Подключение Google Drive для защиты от сброса сессии
print("[*] Подключение к Google Drive...")
drive.mount('/content/drive')

# Определение путей в Google Drive
DRIVE_PROJECT_DIR = '/content/drive/MyDrive/Tesis_EPP'
os.makedirs(DRIVE_PROJECT_DIR, exist_ok=True)
csv_results_path = os.path.join(DRIVE_PROJECT_DIR, 'final_metrics_table.csv')

# Определение аппаратного ускорителя
device = 0 if torch.cuda.is_available() else "cpu"
print(f"[*] Аппаратный ускоритель: {'GPU: ' + torch.cuda.get_device_name(0) if device == 0 else 'CPU'}")

# Глобальные гиперпараметры (20 эпох для баланса времени и точности)
EPOCHS_MAIN = 20
EPOCHS_STAB = 10
IMGSZ = 640
BATCH_SIZE = 16
SEED_LIST = [42, 123, 777]
```

```python
# ---------------------------------------------------------
# 3. ЗАГРУЗКА И ПОДГОТОВКА ДАННЫХ
# ---------------------------------------------------------
print("\n[*] Загрузка датасета из KaggleHub...")
path = kagglehub.dataset_download("snehilsanyal/construction-site-safety-image-dataset-roboflow")
dataset_root = os.path.join(path, "css-data")

train_path = os.path.join(dataset_root, "train/images")
val_path = os.path.join(dataset_root, "valid/images")
test_path = os.path.join(dataset_root, "test/images")

config_data = {
    'train': train_path,
    'val': val_path,
    'test': test_path,
    'nc': 10,
    'names': ['Hardhat', 'Mask', 'NO-Hardhat', 'NO-Mask', 'NO-Safety Vest',
              'Person', 'Safety Cone', 'Safety Vest', 'machinery', 'vehicle']
}

yaml_path = "/content/safety_research.yaml"
with open(yaml_path, 'w') as f:
    yaml.dump(config_data, f)
print(f"[+] Аудит данных: Обучение ({len(os.listdir(train_path))}), Тест ({len(os.listdir(test_path))})")
```

```python
# ---------------------------------------------------------
# 4. ОСНОВНОЙ ЭКСПЕРИМЕНТ (СОХРАНЕНИЕ КАЖДОГО ШАГА НА ДИСК)
# ---------------------------------------------------------
models_to_compare = [
    {'id': 'YOLOv8_Nano',   'weights': 'yolov8n.pt',   'type': 'YOLO'},
    {'id': 'YOLOv8_Medium', 'weights': 'yolov8m.pt',   'type': 'YOLO'},
    {'id': 'YOLOv10_Nano',  'weights': 'yolov10n.pt',  'type': 'YOLO'},
    {'id': 'RT_DETR_L',     'weights': 'rtdetr-l.pt',  'type': 'RTDETR'}
]

# Восстановление прогресса, если часть моделей уже была обучена ранее
if os.path.exists(csv_results_path):
    print(f"[+] Найдены сохраненные ранее результаты. Загрузка из Google Drive...")
    df_existing = pd.read_csv(csv_results_path)
    research_logs = df_existing.to_dict('records')
    completed_models = df_existing['Алгоритм'].tolist()
else:
    research_logs = []
    completed_models = []

for exp in models_to_compare:
    name = exp['id']
    if name in completed_models:
        print(f"[#] Модель {name} уже обучена. Пропуск...")
        continue

    print(f"\n" + "="*50)
    print(f"СТАРТ ЭКСПЕРИМЕНТА: {name}")
    print("="*50)

    model = RTDETR(exp['weights']) if exp['type'] == 'RTDETR' else YOLO(exp['weights'])
    t_start = time.time()

    current_batch = 8 if exp['type'] == 'RTDETR' else BATCH_SIZE

    # Обучение (Сохраняем метеданные и веса прямо в Google Drive)
    model.train(
        data=yaml_path, epochs=EPOCHS_MAIN, imgsz=IMGSZ, batch=current_batch,
        device=device, project=DRIVE_PROJECT_DIR, name=name, exist_ok=True, verbose=False
    )
    train_time_min = (time.time() - t_start) / 60

    print(f"[*] Инференс и валидация {name}...")
    best_model_path = model.trainer.best
    results = model.val(data=yaml_path, split='test', imgsz=IMGSZ, verbose=False)

    # Замер скорости FPS
    test_image_path = os.path.join(test_path, os.listdir(test_path)[0])
    _ = model.predict(test_image_path, conf=0.25, verbose=False) # Прогрев GPU
    t0 = time.time()
    _ = model.predict(test_image_path, conf=0.25, verbose=False)
    t1 = time.time()
    fps = 1 / (t1 - t0) if (t1 - t0) > 0 else 0

    # Добавление лога в список
    row = {
        'Алгоритм': name,
        'mAP@0.5': round(results.box.map50, 3),
        'Precision': round(results.box.mp, 3),
        'Recall': round(results.box.mr, 3),
        'FPS': round(fps, 1),
        'Train_Time_min': round(train_time_min, 1),
        'Params_M': round(sum(p.numel() for p in model.model.parameters()) / 1e6, 2),
        'Size_MB': round(os.path.getsize(best_model_path) / (1024 * 1024), 2)
    }
    research_logs.append(row)

    # Мгновенная запись на Google Drive для защиты данных (Метаданные результатов)
    pd.DataFrame(research_logs).to_csv(csv_results_path, index=False)
    print(f"[+] Результаты для {name} сохранены в Google Drive.")

df_results = pd.DataFrame(research_logs)
print("\n" + "="*50)
print(" ИТОГОВАЯ ТАБЛИЦА СРАВНЕНИЯ АРХИТЕКТУР")
print("="*50)
display(df_results)
```

```python
# ==============================================================================
# БЛОК 5: Оценка статистической стабильности (АВТОНОМНАЯ ВЕРСИЯ ПОСЛЕ СБОЯ)
# ==============================================================================
import os
import torch
import pandas as pd
import yaml
import kagglehub
from ultralytics import YOLO
from google.colab import drive

# 1. Восстанавливаем подключение к Google Drive
if not os.path.exists('/content/drive/MyDrive'):
    print("[*] Переподключение к Google Drive...")
    drive.mount('/content/drive')

DRIVE_PROJECT_DIR = '/content/drive/MyDrive/Tesis_EPP'
yaml_path = "/content/safety_research.yaml"

# 2. Быстрое восстановление датасета, если Colab очистил диск
if not os.path.exists(yaml_path):
    print("[*] Восстановление датасета на диске Colab...")
    path = kagglehub.dataset_download("snehilsanyal/construction-site-safety-image-dataset-roboflow")
    dataset_root = os.path.join(path, "css-data")
    config_data = {
        'train': os.path.join(dataset_root, "train/images"),
        'val': os.path.join(dataset_root, "valid/images"),
        'test': os.path.join(dataset_root, "test/images"),
        'nc': 10,
        'names': ['Hardhat', 'Mask', 'NO-Hardhat', 'NO-Mask', 'NO-Safety Vest',
                  'Person', 'Safety Cone', 'Safety Vest', 'machinery', 'vehicle']
    }
    with open(yaml_path, 'w') as f:
        yaml.dump(config_data, f)

# 3. Восстанавливаем забытые переменные
IMGSZ = 640
device = 0 if torch.cuda.is_available() else "cpu"
SEED_LIST = [42, 123, 777]
EPOCHS_STAB = 10

print("\n[*] Оценка статистической стабильности YOLOv8_Nano (Доверительные интервалы)...")
stability_results = []

# 4. Запуск эксперимента
for s in SEED_LIST:
    print(f"    - Обучение с seed={s}...")
    model_s = YOLO('yolov8n.pt')
    model_s.train(
        data=yaml_path, epochs=EPOCHS_STAB, imgsz=IMGSZ, seed=s,
        device=device, project=DRIVE_PROJECT_DIR, name=f'stability_{s}', exist_ok=True, verbose=False
    )
    # Валидация
    m_s = model_s.val(data=yaml_path, split='test', verbose=False)
    stability_results.append(m_s.box.map50)

# 5. Сохранение метаданных в Google Drive
stability_df = pd.DataFrame({'Seed': SEED_LIST, 'mAP50': stability_results})
stability_df.to_csv(os.path.join(DRIVE_PROJECT_DIR, 'stability_metrics.csv'), index=False)

print(f"\n[*] Результаты mAP@0.5 при разных seed: {[round(res, 3) for res in stability_results]}")
print(f"[*] Среднее стандартное отклонение (σ): {pd.Series(stability_results).std():.4f}")
```

```python
# ==============================================================================
# БЛОК 6: Визуализация и экспорт графиков (АВТОНОМНАЯ ВЕРСИЯ ПОСЛЕ СБОЯ)
# ==============================================================================
import os
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

# 1. Восстанавливаем подключение к Google Drive
if not os.path.exists('/content/drive/MyDrive'):
    print("[*] Переподключение к Google Drive...")
    drive.mount('/content/drive')

DRIVE_PROJECT_DIR = '/content/drive/MyDrive/Tesis_EPP'
csv_results_path = os.path.join(DRIVE_PROJECT_DIR, 'final_metrics_table.csv')

# 2. Автоматическое восстановление таблицы df_results из Google Drive
if os.path.exists(csv_results_path):
    print("[+] Найден сохраненный файл с метриками в Google Drive. Загрузка...")
    df_results = pd.read_csv(csv_results_path)


# 3. Генерация аналитических графиков
print("\n[*]  аналитические графики...")
plt.figure(figsize=(14, 6))
df = df_results.copy()

# График 1: Точность vs Скорость
plt.subplot(1, 2, 1)
plt.grid(True, linestyle='--', alpha=0.7)
plt.scatter(df['FPS'], df['mAP@0.5'], s=150, color='#0b57d0', edgecolors='black', zorder=3)

for i, txt in enumerate(df['Алгоритм']):
    plt.annotate(txt, (df['FPS'][i], df['mAP@0.5'][i]), xytext=(5, 5), textcoords='offset points', fontsize=9)

plt.title('Анализ производительности: Точность vs FPS', fontsize=12, fontweight='bold')
plt.xlabel('Кадры в секунду (FPS)', fontsize=10)
plt.ylabel('Средняя точность ($mAP@0.5$)', fontsize=10)

# График 2: Точность vs Сложность
plt.subplot(1, 2, 2)
plt.grid(True, linestyle='--', alpha=0.7)
plt.scatter(df['Params_M'], df['mAP@0.5'], s=150, color='#0b57d0', edgecolors='black', zorder=3)

for i, txt in enumerate(df['Алгоритм']):
    plt.annotate(txt, (df['Params_M'][i], df['mAP@0.5'][i]), xytext=(5, 5), textcoords='offset points', fontsize=9)

plt.title('Анализ сложности: Точность vs Параметры', fontsize=12, fontweight='bold')
plt.xlabel('Параметры (Миллионы)', fontsize=10)
plt.ylabel('Средняя точность ($mAP@0.5$)', fontsize=10)

plt.tight_layout()
graph_save_path = os.path.join(DRIVE_PROJECT_DIR, 'research_analysis_plots.png')
plt.savefig(graph_save_path, dpi=300, bbox_inches='tight')
print(f"[+] Эксперимент полностью завершен! Графики и таблицы сохранены в Google Drive: {DRIVE_PROJECT_DIR}")
plt.show()
```
